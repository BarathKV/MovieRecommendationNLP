# Process the vector representation extracted from Word2Vec model

## Save each vector as separate dataframe objects using pickle

```python
import pandas as pd

data = pd.read_csv('../Dataset/combined_with_vectors.csv')
data.head()
```

```python
import pickle
import numpy as np

# Function to process and reshape vector columns
def process_vector_column(column_name, data):
    # Step 1: Clean the string in each row of the specified column
    data[column_name] = data[column_name].apply(
        lambda x: np.array([float(i) for i in x.replace("[", "").replace("]", "").replace("\n", " ").strip().split()])
    )

    # Step 2: Reshape each vector to have 1 row and N columns
    data[column_name] = data[column_name].apply(lambda x: x.reshape(1, -1))

    # Optional: Convert to pandas DataFrame (if required to work with later)
    data[column_name] = data[column_name].apply(lambda x: pd.DataFrame(x))

    return data

# Apply to all relevant columns
data = process_vector_column('keywords_vector', data)
data = process_vector_column('genres_vector', data)
data = process_vector_column('overview_vector', data)

# Now the columns in 'data' have been updated with reshaped NumPy arrays
data.head()
```

```python
from scipy.sparse import csr_matrix, vstack
def create_sparse_matrix(data):
    # Convert each vector in the specified column to a sparse matrix
    sparse_matrices = [csr_matrix(vec) for vec in data]

    # Stack them all vertically
    stacked_matrix = vstack(sparse_matrices)

    return stacked_matrix
```

```python
with open('../Matrices/w2v/combined_vectors.pkl', 'rb') as f:
    combined_vector = pickle.load(f)

combined_vector.head()
```

```python
keyword_matrix = create_sparse_matrix(combined_vector['keywords_vector'])
with open('../Matrices/w2v/keyword_mtx.pkl', 'wb') as f:
    pickle.dump(keyword_matrix, f)
keyword_matrix.shape
```

```python
genres_matrix = create_sparse_matrix(combined_vector['genres_vector'])
with open('../Matrices/w2v/genres_mtx.pkl', 'wb') as f:
    pickle.dump(genres_matrix, f)
genres_matrix.shape
```

```python
overview_matrix = create_sparse_matrix(combined_vector['overview_vector'])
with open('../Matrices/w2v/overview_mtx.pkl', 'wb') as f:
    pickle.dump(overview_matrix, f)
overview_matrix.shape
```

```python
from scipy.sparse import hstack

# Concatenate sparse matrices
combined_matrix = hstack((keyword_matrix, genres_matrix, overview_matrix))
with open('../Matrices/w2v/combined_mtx.pkl', 'wb') as f:
    pickle.dump(combined_matrix, f)
combined_matrix.shape
```

```python
from sklearn.preprocessing import normalize

# Normalize the combined sparse matrix
normalized_matrix = normalize(combined_matrix)
with open('../Matrices/w2v/normalized_mtx.pkl', 'wb') as f:
    pickle.dump(normalized_matrix, f)
normalized_matrix.shape
```

```python
from sklearn.decomposition import TruncatedSVD

# Reduce dimensions using Truncated SVD
svd = TruncatedSVD(n_components=100, random_state=42)
reduced_matrix = svd.fit_transform(normalized_matrix)
with open('../Matrices/w2v/reduced_mtx.pkl', 'wb') as f:
    pickle.dump(reduced_matrix, f)
reduced_matrix.shape
```
